```python
# ------------------------------------------------------------
# STEP 3: Enter your Mistral API key
# ------------------------------------------------------------

# Load API key from Colab Secrets
import os
from google.colab import userdata
from mistralai.client import Mistral
os.environ["MISTRAL_API_KEY"] = userdata.get("MISTRAL_API_KEY")

# Initialize the new client
client = Mistral(api_key=os.environ["MISTRAL_API_KEY"])
```

```python
# ------------------------------------------------------------
# 2. Question
# ------------------------------------------------------------

question = """
Answer in one short sentence:
What is artificial intelligence?
"""

# ------------------------------------------------------------
# 3. Human reference answer
# ------------------------------------------------------------

reference_answer = (
    "Artificial intelligence is the ability of machines "
    "to perform tasks that normally require human intelligence."
)

# ------------------------------------------------------------
# 4. Call Mistral
# ------------------------------------------------------------

response = client.chat.complete(
    model="mistral-small-latest",
    messages=[
        {
            "role": "user",
            "content": question
        }
    ]
)

# ------------------------------------------------------------
# 5. Extract answer
# ------------------------------------------------------------

mistral_answer = response.choices[0].message.content.strip()

# ------------------------------------------------------------
# 6. Calculate BERTScore
# ------------------------------------------------------------

P, R, F1 = score(
    [mistral_answer],
    [reference_answer],
    lang="en",
    model_type="roberta-large",
    verbose=False
)

# ------------------------------------------------------------
# 7. Display results
# ------------------------------------------------------------

print("\n======================================")
print("     MISTRAL BERTSCORE EVALUATION")
print("======================================")

print("\nQUESTION:")
print(question.strip())

print("\nREFERENCE ANSWER:")
print(reference_answer)

print("\nMISTRAL ANSWER:")
print(mistral_answer)

print("\n======================================")
print("            BERTSCORE")
print("======================================")

print("Precision :", round(P.item(), 4))
print("Recall    :", round(R.item(), 4))
print("F1 Score  :", round(F1.item(), 4))
```
