```python
# Install if required:
# pip install transformers torch

from transformers import AutoModelForCausalLM, AutoConfig

model_name = "distilgpt2"

# ------------------------------------
# 1. Load model configuration
# ------------------------------------
config = AutoConfig.from_pretrained(model_name)

print("\n========== MODEL HYPERPARAMETERS ==========\n")

print("Model type:", config.model_type)
print("Vocabulary size:", config.vocab_size)
print("Hidden size:", config.n_embd)
print("Number of transformer layers:", config.n_layer)
print("Number of attention heads:", config.n_head)
print("Maximum context length:", config.n_positions)

# ------------------------------------
# 2. Load actual model
# ------------------------------------
model = AutoModelForCausalLM.from_pretrained(model_name)

# ------------------------------------
# 3. Count total parameters
# ------------------------------------
total_params = sum(p.numel() for p in model.parameters())

print("\n========== MODEL PARAMETERS ==========\n")

print("Total parameters:", total_params)
print("Total parameters in millions:",
      round(total_params / 1_000_000, 2), "Million")

# ------------------------------------
# 4. Count trainable parameters
# ------------------------------------
trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print("Trainable parameters:", trainable_params)

# ------------------------------------
# 5. Display individual parameter tensors
# ------------------------------------
print("\n========== SAMPLE PARAMETER LAYERS ==========\n")

for name, parameter in list(model.named_parameters())[:10]:

    print("Parameter name:", name)
    print("Shape:", parameter.shape)
    print("Number of values:", parameter.numel())
    print()
```
