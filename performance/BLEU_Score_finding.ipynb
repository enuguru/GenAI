```python
import os
from google import genai
from google.colab import userdata

#API_KEY = put the new api key here created latest
api_key = userdata.get("GOOGLE_API_KEY")
client = genai.Client(api_key=api_key)
os.environ["GOOGLE_API_KEY"] = api_key
# ------------------------------------------------------------
# 1. Enter your Gemini API key
# ------------------------------------------------------------

#GEMINI_API_KEY = input("Enter your Gemini API Key: ")

#client = genai.Client(api_key=GEMINI_API_KEY)
```

```python


# ------------------------------------------------------------
# 2. Question given to the commercial LLM
# ------------------------------------------------------------

question = """
Answer in ONE short sentence:
What is the capital of France?
"""

# ------------------------------------------------------------
# 3. Human-created reference answer
# ------------------------------------------------------------

reference_answer = "The capital of France is Paris."

# ------------------------------------------------------------
# 4. Ask the actual Gemini LLM
# ------------------------------------------------------------

response = client.models.generate_content(
    model="gemini-3.7-flash",
    contents=question
)

llm_answer = response.text.strip()

# ------------------------------------------------------------
# 5. Tokenize reference and LLM answer
# ------------------------------------------------------------

reference_tokens = reference_answer.lower().split()
candidate_tokens = llm_answer.lower().split()

# ------------------------------------------------------------
# 6. Calculate BLEU
# ------------------------------------------------------------

smooth = SmoothingFunction().method1

bleu = sentence_bleu(
    [reference_tokens],
    candidate_tokens,
    smoothing_function=smooth
)

# ------------------------------------------------------------
# 7. Display results
# ------------------------------------------------------------

print("\nQUESTION:")
print(question)

print("\nHUMAN REFERENCE:")
print(reference_answer)

print("\nGEMINI ANSWER:")
print(llm_answer)

print("\nBLEU SCORE:")
print(round(bleu, 4))
```
