```python
# ------------------------------------------------------------
# STEP 2: Imports
# ------------------------------------------------------------

from google import genai
from rouge_score import rouge_scorer
import getpass


# ------------------------------------------------------------
# STEP 3: Enter Gemini API Key
# ------------------------------------------------------------
import os
from google import genai
from google.colab import userdata

#API_KEY = put the new api key here created latest
api_key = userdata.get("GOOGLE_API_KEY")
#client = genai.Client(api_key=api_key)
os.environ["GOOGLE_API_KEY"] = api_key
#api_key = getpass.getpass("Enter your Gemini API Key: ")

client = genai.Client(api_key=api_key)
```

```python
# ------------------------------------------------------------
# STEP 4: Question
# ------------------------------------------------------------

question = """
Answer in one short sentence:
What is artificial intelligence?
"""


# ------------------------------------------------------------
# STEP 5: Human Reference Answer
# ------------------------------------------------------------

reference_answer = (
    "Artificial intelligence is the ability of machines "
    "to perform tasks that normally require human intelligence."
)


# ------------------------------------------------------------
# STEP 6: Ask Gemini
# ------------------------------------------------------------

response = client.models.generate_content(
    model="gemini-3.7-flash",
    contents=question
)


# ------------------------------------------------------------
# STEP 7: Extract Gemini response
# ------------------------------------------------------------

gemini_answer = response.text.strip()


# ------------------------------------------------------------
# STEP 8: Create ROUGE evaluator
# ------------------------------------------------------------

scorer = rouge_scorer.RougeScorer(
    ['rouge1', 'rouge2', 'rougeL'],
    use_stemmer=True
)


# ------------------------------------------------------------
# STEP 9: Calculate ROUGE
# ------------------------------------------------------------

scores = scorer.score(
    reference_answer,
    gemini_answer
)


# ------------------------------------------------------------
# STEP 10: Display Gemini Answer
# ------------------------------------------------------------

print("\n==========================================")
print("       GEMINI LLM ROUGE EVALUATION")
print("==========================================")

print("\nQUESTION:")
print(question.strip())

print("\nREFERENCE ANSWER:")
print(reference_answer)

print("\nGEMINI ANSWER:")
print(gemini_answer)


# ------------------------------------------------------------
# STEP 11: Display ROUGE Scores
# ------------------------------------------------------------

print("\n==========================================")
print("              ROUGE SCORES")
print("==========================================")

for metric, score in scores.items():

    print(f"\n{metric.upper()}")

    print("Precision :", round(score.precision, 4))
    print("Recall    :", round(score.recall, 4))
    print("F1 Score  :", round(score.fmeasure, 4))
```
