```python
"""
Dynamic Subagent Demonstration

Concept:
--------
Main Agent
    ↓
Analyzes User Request
    ↓
Creates ONLY required Subagents
    ↓
Delegates Work
    ↓
Collects Results
    ↓
Destroys Dynamic Subagents
"""

import time


# -------------------------------------------------------
# Base Dynamic Subagent
# -------------------------------------------------------
class DynamicSubAgent:

    def __init__(self, name, expertise):
        self.name = name
        self.expertise = expertise

        print(f"\n[CREATE] {self.name} created.")
        print(f"         Expertise : {self.expertise}")

    def execute(self, task):

        print(f"\n{self.name} is working...")
        time.sleep(1)

        return f"{self.name} completed: {task}"

    def __del__(self):
        print(f"[DESTROY] {self.name} destroyed.")


# -------------------------------------------------------
# Dynamic Agent Factory
# -------------------------------------------------------
class AgentFactory:

    @staticmethod
    def create(agent_type):

        if agent_type == "research":
            return DynamicSubAgent(
                "Research Expert",
                "Internet Research"
            )

        elif agent_type == "coding":
            return DynamicSubAgent(
                "Coding Expert",
                "Python Development"
            )

        elif agent_type == "finance":
            return DynamicSubAgent(
                "Finance Expert",
                "Financial Analysis"
            )

        elif agent_type == "presentation":
            return DynamicSubAgent(
                "Presentation Expert",
                "PowerPoint Generation"
            )

        else:
            return None


# -------------------------------------------------------
# Main Agent
# -------------------------------------------------------
class MainAgent:

    def understand_request(self, request):

        print("\nMain Agent analyzing request...")

        required = []

        request = request.lower()

        if "research" in request:
            required.append("research")

        if "python" in request:
            required.append("coding")

        if "finance" in request:
            required.append("finance")

        if "presentation" in request:
            required.append("presentation")

        return required

    def solve(self, request):

        print("=" * 70)
        print("USER REQUEST")
        print(request)

        required_agents = self.understand_request(request)

        print("\nRequired Dynamic Subagents:")
        print(required_agents)

        results = []

        print("\nCreating Dynamic Subagents...\n")

        for agent_type in required_agents:

            # Create only when needed
            agent = AgentFactory.create(agent_type)

            result = agent.execute(
                f"Handle {agent_type} portion"
            )

            results.append(result)

            # Destroy immediately after work
            del agent

        print("\nMain Agent combining results...\n")

        print("=" * 70)
        print("FINAL RESPONSE\n")

        for r in results:
            print("•", r)

        print("\nTask Completed")
        print("=" * 70)


# -------------------------------------------------------
# Driver
# -------------------------------------------------------

if __name__ == "__main__":

    user_request = """
    Research Agentic AI,
    generate Python code,
    perform finance analysis,
    and create presentation.
    """

    main_agent = MainAgent()
    main_agent.solve(user_request)
```
