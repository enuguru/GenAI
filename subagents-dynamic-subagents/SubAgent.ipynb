```python
from dataclasses import dataclass
from typing import List


@dataclass
class AgentResult:
    agent_name: str
    output: str


class SubAgent:
    """Base class for all subagents."""

    def __init__(self, name: str, responsibility: str):
        self.name = name
        self.responsibility = responsibility

    def execute(self, task: str) -> AgentResult:
        raise NotImplementedError("Subclasses must implement execute().")


class ResearchSubAgent(SubAgent):
    def __init__(self):
        super().__init__(
            name="Research Subagent",
            responsibility="Collect relevant information"
        )

    def execute(self, task: str) -> AgentResult:
        print(f"{self.name} received task: {task}")

        # Simulated research result
        result = (
            "Agentic AI systems use planning, tool calling, memory, "
            "reflection, and multi-agent coordination."
        )

        return AgentResult(self.name, result)


class CodingSubAgent(SubAgent):
    def __init__(self):
        super().__init__(
            name="Coding Subagent",
            responsibility="Generate technical examples"
        )

    def execute(self, task: str) -> AgentResult:
        print(f"{self.name} received task: {task}")

        # Simulated code-generation result
        result = (
            "Example: A parent agent can call specialized Python classes "
            "to perform research, analysis, and report generation."
        )

        return AgentResult(self.name, result)


class WritingSubAgent(SubAgent):
    def __init__(self):
        super().__init__(
            name="Writing Subagent",
            responsibility="Create a clear final explanation"
        )

    def execute(self, task: str) -> AgentResult:
        print(f"{self.name} received task: {task}")

        result = (
            "Subagents improve modularity by allowing each agent to focus "
            "on a specific responsibility."
        )

        return AgentResult(self.name, result)


class MainAgent:
    """Main agent that plans, delegates, and combines results."""

    def __init__(self):
        self.subagents = {
            "research": ResearchSubAgent(),
            "coding": CodingSubAgent(),
            "writing": WritingSubAgent()
        }

    def create_plan(self, user_request: str) -> List[tuple]:
        print("\nMain Agent is creating a plan...")

        return [
            (
                "research",
                f"Research the main concepts related to: {user_request}"
            ),
            (
                "coding",
                f"Prepare a technical example for: {user_request}"
            ),
            (
                "writing",
                f"Create a simple explanation for: {user_request}"
            )
        ]

    def delegate_tasks(self, plan: List[tuple]) -> List[AgentResult]:
        results = []

        print("\nMain Agent is delegating tasks...\n")

        for agent_key, task in plan:
            subagent = self.subagents[agent_key]
            result = subagent.execute(task)
            results.append(result)

        return results

    def combine_results(
        self,
        user_request: str,
        results: List[AgentResult]
    ) -> str:
        print("\nMain Agent is combining the results...")

        sections = [f"Final response for: {user_request}\n"]

        for result in results:
            sections.append(
                f"{result.agent_name}:\n{result.output}\n"
            )

        sections.append(
            "Main Agent Summary:\n"
            "The main agent owns the overall objective, while each "
            "subagent completes a specialized part of the work."
        )

        return "\n".join(sections)

    def run(self, user_request: str) -> str:
        print(f"User request: {user_request}")

        plan = self.create_plan(user_request)
        results = self.delegate_tasks(plan)
        final_response = self.combine_results(user_request, results)

        return final_response


def main():
    main_agent = MainAgent()

    user_request = (
        "Explain how subagents are used in an Agentic AI system"
    )

    final_response = main_agent.run(user_request)

    print("\n" + "=" * 70)
    print(final_response)
    print("=" * 70)


if __name__ == "__main__":
    main()
```
