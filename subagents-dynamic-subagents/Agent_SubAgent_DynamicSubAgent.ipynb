```python
"""
Demonstration of:
1. Main Agent
2. Static Subagents
3. Dynamic Subagents

No external libraries are required.
Run with:

    python agent_demo.py
"""

from dataclasses import dataclass
from typing import Dict, List, Optional


# ============================================================
# Result returned by every subagent
# ============================================================

@dataclass
class AgentResult:
    agent_name: str
    agent_type: str
    task: str
    output: str


# ============================================================
# Base class for all subagents
# ============================================================

class BaseSubAgent:
    """Common base class for static and dynamic subagents."""

    def __init__(self, name: str, expertise: str):
        self.name = name
        self.expertise = expertise

    def execute(self, task: str) -> AgentResult:
        raise NotImplementedError(
            "Every subagent must implement the execute() method."
        )


# ============================================================
# STATIC SUBAGENTS
#
# These are created when the MainAgent starts.
# They remain available throughout the application.
# ============================================================

class ResearchSubAgent(BaseSubAgent):

    def __init__(self):
        super().__init__(
            name="Research Subagent",
            expertise="Collecting and organizing information"
        )

    def execute(self, task: str) -> AgentResult:
        print(f"\n[STATIC SUBAGENT] {self.name}")
        print(f"Task: {task}")

        output = (
            "Research completed. The topic involves a main agent that "
            "coordinates specialized subagents."
        )

        return AgentResult(
            agent_name=self.name,
            agent_type="Static Subagent",
            task=task,
            output=output
        )


class WritingSubAgent(BaseSubAgent):

    def __init__(self):
        super().__init__(
            name="Writing Subagent",
            expertise="Creating clear explanations and summaries"
        )

    def execute(self, task: str) -> AgentResult:
        print(f"\n[STATIC SUBAGENT] {self.name}")
        print(f"Task: {task}")

        output = (
            "A structured explanation was prepared using simple "
            "language and clear sections."
        )

        return AgentResult(
            agent_name=self.name,
            agent_type="Static Subagent",
            task=task,
            output=output
        )


# ============================================================
# DYNAMIC SUBAGENTS
#
# These are not permanently stored inside MainAgent.
# They are created only when a request requires them.
# ============================================================

class CodingDynamicSubAgent(BaseSubAgent):

    def __init__(self):
        super().__init__(
            name="Dynamic Coding Specialist",
            expertise="Python programming"
        )

        print(f"\n[CREATED DYNAMICALLY] {self.name}")

    def execute(self, task: str) -> AgentResult:
        print(f"[DYNAMIC SUBAGENT] {self.name}")
        print(f"Task: {task}")

        output = (
            "Python code was generated using classes, delegation, "
            "runtime agent creation, and result aggregation."
        )

        return AgentResult(
            agent_name=self.name,
            agent_type="Dynamic Subagent",
            task=task,
            output=output
        )


class SecurityDynamicSubAgent(BaseSubAgent):

    def __init__(self):
        super().__init__(
            name="Dynamic Security Specialist",
            expertise="Security review and risk analysis"
        )

        print(f"\n[CREATED DYNAMICALLY] {self.name}")

    def execute(self, task: str) -> AgentResult:
        print(f"[DYNAMIC SUBAGENT] {self.name}")
        print(f"Task: {task}")

        output = (
            "Security review completed. Input validation, access control, "
            "logging, and safe tool permissions were recommended."
        )

        return AgentResult(
            agent_name=self.name,
            agent_type="Dynamic Subagent",
            task=task,
            output=output
        )


class DataAnalysisDynamicSubAgent(BaseSubAgent):

    def __init__(self):
        super().__init__(
            name="Dynamic Data Analysis Specialist",
            expertise="Data processing and statistical analysis"
        )

        print(f"\n[CREATED DYNAMICALLY] {self.name}")

    def execute(self, task: str) -> AgentResult:
        print(f"[DYNAMIC SUBAGENT] {self.name}")
        print(f"Task: {task}")

        output = (
            "The data-analysis requirement was identified and an "
            "appropriate analysis workflow was proposed."
        )

        return AgentResult(
            agent_name=self.name,
            agent_type="Dynamic Subagent",
            task=task,
            output=output
        )


# ============================================================
# DYNAMIC SUBAGENT FACTORY
#
# The factory creates the requested specialist at runtime.
# ============================================================

class DynamicSubAgentFactory:

    @staticmethod
    def create(agent_type: str) -> Optional[BaseSubAgent]:

        dynamic_agents = {
            "coding": CodingDynamicSubAgent,
            "security": SecurityDynamicSubAgent,
            "data_analysis": DataAnalysisDynamicSubAgent
        }

        agent_class = dynamic_agents.get(agent_type)

        if agent_class is None:
            return None

        return agent_class()


# ============================================================
# MAIN AGENT
#
# This is the top-level agent.
# It owns the user's overall goal.
# ============================================================

class MainAgent:

    def __init__(self):
        self.name = "Main Orchestrator Agent"

        # Static subagents are created during initialization.
        self.static_subagents: Dict[str, BaseSubAgent] = {
            "research": ResearchSubAgent(),
            "writing": WritingSubAgent()
        }

        print("=" * 70)
        print(f"{self.name} initialized.")
        print("Static subagents are ready:")
        print("  - Research Subagent")
        print("  - Writing Subagent")
        print("=" * 70)

    # --------------------------------------------------------
    # Main agent analyzes the user's request.
    # --------------------------------------------------------

    def analyze_request(self, user_request: str) -> List[str]:
        request = user_request.lower()

        required_dynamic_agents = []

        if any(word in request for word in ["python", "code", "program"]):
            required_dynamic_agents.append("coding")

        if any(word in request for word in ["security", "secure", "risk"]):
            required_dynamic_agents.append("security")

        if any(
            phrase in request
            for phrase in ["data analysis", "statistics", "dataset"]
        ):
            required_dynamic_agents.append("data_analysis")

        return required_dynamic_agents

    # --------------------------------------------------------
    # Main agent creates a task plan.
    # --------------------------------------------------------

    def create_plan(
        self,
        user_request: str,
        dynamic_agent_types: List[str]
    ) -> List[dict]:

        plan = [
            {
                "agent_category": "static",
                "agent_type": "research",
                "task": f"Research the concepts required for: {user_request}"
            },
            {
                "agent_category": "static",
                "agent_type": "writing",
                "task": f"Prepare a clear explanation for: {user_request}"
            }
        ]

        for agent_type in dynamic_agent_types:
            plan.append(
                {
                    "agent_category": "dynamic",
                    "agent_type": agent_type,
                    "task": (
                        f"Handle the specialized {agent_type} requirement "
                        f"for: {user_request}"
                    )
                }
            )

        return plan

    # --------------------------------------------------------
    # Execute one static subagent.
    # --------------------------------------------------------

    def run_static_subagent(
        self,
        agent_type: str,
        task: str
    ) -> AgentResult:

        subagent = self.static_subagents[agent_type]
        return subagent.execute(task)

    # --------------------------------------------------------
    # Create and execute one dynamic subagent.
    # --------------------------------------------------------

    def run_dynamic_subagent(
        self,
        agent_type: str,
        task: str
    ) -> AgentResult:

        dynamic_subagent = DynamicSubAgentFactory.create(agent_type)

        if dynamic_subagent is None:
            raise ValueError(
                f"No dynamic subagent is registered for '{agent_type}'."
            )

        result = dynamic_subagent.execute(task)

        print(f"[RELEASED] {dynamic_subagent.name}")

        # The dynamic subagent is not stored permanently.
        # After this method finishes, it becomes eligible for cleanup.
        del dynamic_subagent

        return result

    # --------------------------------------------------------
    # Main agent combines all subagent results.
    # --------------------------------------------------------

    def aggregate_results(
        self,
        user_request: str,
        results: List[AgentResult]
    ) -> str:

        report = [
            "\n" + "=" * 70,
            "FINAL RESPONSE FROM MAIN AGENT",
            "=" * 70,
            f"\nOriginal request:\n{user_request}\n"
        ]

        for number, result in enumerate(results, start=1):
            report.extend(
                [
                    f"{number}. {result.agent_name}",
                    f"   Type   : {result.agent_type}",
                    f"   Task   : {result.task}",
                    f"   Output : {result.output}",
                    ""
                ]
            )

        report.extend(
            [
                "Main Agent Summary:",
                (
                    "The main agent owned the complete goal, delegated "
                    "regular work to predefined static subagents, created "
                    "specialized dynamic subagents only when required, and "
                    "combined all outputs into one final response."
                ),
                "=" * 70
            ]
        )

        return "\n".join(report)

    # --------------------------------------------------------
    # Complete agent workflow.
    # --------------------------------------------------------

    def run(self, user_request: str) -> str:

        print(f"\nUSER REQUEST:\n{user_request}")

        print("\n[MAIN AGENT] Analyzing the request...")

        dynamic_agent_types = self.analyze_request(user_request)

        if dynamic_agent_types:
            print(
                "Dynamic specialists required:",
                ", ".join(dynamic_agent_types)
            )
        else:
            print("No dynamic specialists are required.")

        print("\n[MAIN AGENT] Creating execution plan...")

        plan = self.create_plan(
            user_request,
            dynamic_agent_types
        )

        results: List[AgentResult] = []

        print("\n[MAIN AGENT] Delegating tasks...")

        for plan_item in plan:
            category = plan_item["agent_category"]
            agent_type = plan_item["agent_type"]
            task = plan_item["task"]

            if category == "static":
                result = self.run_static_subagent(
                    agent_type,
                    task
                )
            else:
                result = self.run_dynamic_subagent(
                    agent_type,
                    task
                )

            results.append(result)

        print("\n[MAIN AGENT] Aggregating all results...")

        return self.aggregate_results(
            user_request,
            results
        )


# ============================================================
# Program entry point
# ============================================================

def main():

    main_agent = MainAgent()

    user_request = (
        "Explain agentic AI, create a Python program, "
        "and perform a security review."
    )

    final_response = main_agent.run(user_request)

    print(final_response)


if __name__ == "__main__":
    main()
```
