```python
# ==========================================
# MULTI-AGENT COORDINATION DEMO
# Route + Traffic + Charging + Planner
# ==========================================

# ------------------------------------------
# Route Agent
# ------------------------------------------
class RouteAgent:

    def get_routes(self, source, destination):

        print("\n[ROUTE AGENT]")
        print(f"Received task: Find routes from {source} to {destination}")

        routes = ["Route A", "Route B"]

        print("Output:", routes)

        return routes


# ------------------------------------------
# Traffic Agent
# ------------------------------------------
class TrafficAgent:

    def check_traffic(self, routes):

        print("\n[TRAFFIC AGENT]")
        print("Received routes:", routes)

        traffic_info = {
            "Route A": "Heavy Traffic",
            "Route B": "Light Traffic"
        }

        print("Output:", traffic_info)

        return traffic_info


# ------------------------------------------
# Charging Agent
# ------------------------------------------
class ChargingAgent:

    def check_charging(self, routes):

        print("\n[CHARGING AGENT]")
        print("Received routes:", routes)

        charging_info = {
            "Route A": "No charging station nearby",
            "Route B": "Fast charger available after 180 km"
        }

        print("Output:", charging_info)

        return charging_info


# ------------------------------------------
# Planner Agent
# ------------------------------------------
class PlannerAgent:

    def __init__(self):

        self.route_agent = RouteAgent()
        self.traffic_agent = TrafficAgent()
        self.charging_agent = ChargingAgent()


    def plan_trip(self, source, destination):

        print("\n======================================")
        print("PLANNER AGENT")
        print("======================================")

        print(f"Goal: Travel from {source} to {destination}")


        # ----------------------------------
        # Agent-to-Agent Communication 1
        # Planner -> Route Agent
        # ----------------------------------

        print("\nPlanner Agent -> Route Agent")

        routes = self.route_agent.get_routes(
            source,
            destination
        )


        # ----------------------------------
        # Route Agent output becomes
        # Traffic Agent input
        # ----------------------------------

        print("\nRoute Agent OUTPUT -> Traffic Agent INPUT")

        traffic = self.traffic_agent.check_traffic(
            routes
        )


        # ----------------------------------
        # Route Agent output also becomes
        # Charging Agent input
        # ----------------------------------

        print("\nRoute Agent OUTPUT -> Charging Agent INPUT")

        charging = self.charging_agent.check_charging(
            routes
        )


        # ----------------------------------
        # Traffic + Charging outputs
        # become Planner Agent inputs
        # ----------------------------------

        print("\nTraffic Agent OUTPUT -> Planner Agent INPUT")
        print("Charging Agent OUTPUT -> Planner Agent INPUT")


        print("\n======================================")
        print("PLANNER AGENT ANALYSIS")
        print("======================================")

        for route in routes:

            print("\n", route)

            print("Traffic  :", traffic[route])
            print("Charging :", charging[route])


        # ----------------------------------
        # Planner reasoning
        # ----------------------------------

        if (
            traffic["Route B"] == "Light Traffic"
            and
            "charger" in charging["Route B"].lower()
        ):

            final_route = "Route B"

        else:

            final_route = "Route A"


        print("\n======================================")
        print("FINAL TRIP PLAN")
        print("======================================")

        print("Selected Route:", final_route)

        print(
            "Reason:",
            traffic[final_route],
            "+",
            charging[final_route]
        )


# ==========================================
# RUN THE MULTI-AGENT SYSTEM
# ==========================================

planner = PlannerAgent()

planner.plan_trip(
    "Bengaluru",
    "Chennai"
)
```
