```python
# Simple Agent-to-Agent Communication Example

class Agent:
    def __init__(self, name):
        self.name = name

    def send_message(self, receiver, message):
        print(f"\n{self.name} -> {receiver.name}")
        print("Message:", message)

        # Send message to another agent
        response = receiver.receive_message(self.name, message)

        return response

    def receive_message(self, sender, message):
        print(f"{self.name} received message from {sender}")

        # Simple agent reasoning
        if "traffic" in message.lower():
            response = "Traffic is heavy. Take Route B."
        elif "weather" in message.lower():
            response = "Weather is clear."
        else:
            response = "Task completed."

        print(f"{self.name} -> Response: {response}")

        return response


# --------------------------------
# Create two agents
# --------------------------------

navigation_agent = Agent("Navigation Agent")
traffic_agent = Agent("Traffic Agent")


# --------------------------------
# Agent-to-Agent Communication
# --------------------------------

response = navigation_agent.send_message(
    traffic_agent,
    "Please check traffic on Route A."
)


# --------------------------------
# Navigation Agent uses response
# --------------------------------

print("\nFINAL DECISION")

if "Route B" in response:
    print("Navigation Agent: I will use Route B.")
else:
    print("Navigation Agent: I will continue on Route A.")
```
