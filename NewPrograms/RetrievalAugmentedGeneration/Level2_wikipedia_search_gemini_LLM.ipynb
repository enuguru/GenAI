```python
# ============================================================
# 3. GEMINI SETUP
# ============================================================

import os
from google import genai
from google.colab import userdata

#API_KEY = put the new api key here created latest
api_key = userdata.get("GOOGLE_API_KEY")
client = genai.Client(api_key=api_key)
os.environ["GOOGLE_API_KEY"] = api_key
#import os
#from getpass import getpass

#os.environ["GOOGLE_API_KEY"] = getpass("Enter Gemini API key: ")
```

```python
# ============================================================
# 2. FUNCTION TO SEARCH WIKIPEDIA
# ============================================================

def search_wikipedia(question):
    print("\nSearching Wikipedia...")
    # Search Wikipedia
    search_results = wikipedia.search(
        question,
        results=5
    )
    if not search_results:
        return None, None
    print("\nWikipedia search results:")
    for i, title in enumerate(search_results, 1):
        print(i, title)

    # Use first search result
    selected_title = search_results[0]

    print(
        "\nSelected Wikipedia article:",
        selected_title
    )

    try:
        page = wikipedia.page(
            selected_title,
            auto_suggest=False
        )

        # Get article content
        content = page.content
        return selected_title, content

    except wikipedia.DisambiguationError as e:

        # If ambiguous, select first suggestion
        selected_title = e.options[0]
        page = wikipedia.page(
            selected_title,
            auto_suggest=False
        )
        return selected_title, page.content
    except Exception as e:
        print("Wikipedia error:", e)
        return None, None
```

```python

# ============================================================
# 3. ASK USER QUESTION
# ============================================================

question = input(
    "\nEnter your question: "
)

# ============================================================
# 4. RETRIEVE KNOWLEDGE FROM WIKIPEDIA
# ============================================================

title, wikipedia_text = search_wikipedia(
    question
)
if wikipedia_text is None:
    print(
        "\nCould not retrieve information from Wikipedia."
    )
else:
    # Limit context size for demonstration
    wikipedia_context = wikipedia_text[:12000]
```

```python
    # ========================================================
    # 5. DISPLAY RETRIEVED KNOWLEDGE
    # ========================================================

    print("\n")
    print("=" * 70)
    print("WIKIPEDIA KNOWLEDGE RETRIEVED")
    print("=" * 70)
    print("\nArticle:", title)
    print("\nRetrieved text:\n")
    print(
        wikipedia_context[:2000]
    )
    print("\n...")

    # ========================================================
    # 6. CREATE PROMPT FOR GEMINI
    # ========================================================

    prompt = f"""
You are an AI assistant.
A user has asked the following question:
QUESTION:
{question}
Wikipedia has provided the following information:
WIKIPEDIA ARTICLE:
{title}
WIKIPEDIA CONTENT:
{wikipedia_context}

Using the Wikipedia information above,
answer the user's question clearly and accurately.

Do not invent information that is not supported by
the supplied Wikipedia content.

If the supplied information is insufficient,
say that the Wikipedia information is insufficient.

Give a clear, concise and well-structured answer.
"""

    # ========================================================
    # 7. SEND INFORMATION TO GEMINI
    # ========================================================

    print("\nSending retrieved information to Gemini...")
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )


    # ========================================================
    # 8. DISPLAY FINAL ANSWER
    # ========================================================

    print("\n")
    print("=" * 70)
    print("FINAL GEMINI ANSWER")
    print("=" * 70)
    print()
    print(
        response.text
    )
```
